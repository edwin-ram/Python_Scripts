```python
import pandas as pd

df_estados_ifactura = pd.read_csv("documentos/inputs/Informe Keralty_Exitoso_Tabla.csv", sep=",")

df_estados_facturas = df_estados_ifactura[["NumeroDocumento","ESTADO"]]

print(df_estados_facturas.tail(10))
```

```python
df_logfactura = pd.read_csv('documentos/INFORME LOG FACTURACION ELECTRONICA INT 04092026 55977572.csv', sep = '|')
df_logfactura = df_logfactura.drop_duplicates(subset=['Número de Documento'], keep=False)
df_facturas = df_logfactura [["Número de Documento","Estado"]]

print(df_facturas.tail(5))
print(f'Existen {len(df_facturas["Número de Documento"])}')
```

```python
df_resultante = pd.merge(
    df_facturas, #Desde donde voy a buscar
    df_estados_facturas[['NumeroDocumento','ESTADO']], #buscar en... es como seleccionar la tabla y dar F4
    left_on = 'Número de Documento', #Columna donde esta el valor buscado desde df_facturas
    right_on='NumeroDocumento', #Columna donde se se va a iterar la busqueda en df_estados_facturas
    how='left') #Respeta lo que haya en la tabla principal (df_facturas) es como decir verdadadero en buscarv

print(f'Existen {len(df_resultante["Número de Documento"])}')
```

```python
df_novedades = df_resultante[(df_resultante['Estado']!="AUT-DIAN") & (df_resultante['ESTADO']!="EXITOSO")]
df_rptaokpt = df_resultante[(df_resultante['Estado']!="AUT-DIAN") & (df_resultante['ESTADO']=="EXITOSO")]
print(f'Se encontraron {len(df_novedades["Número de Documento"])} con novedades')
print(f'Se encontraron {len(df_rptaokpt["Número de Documento"])} en estado RPTAOKPT')
```

```python
with open ('outputs/11092026.txt', 'w', encoding='utf-8') as f:
    for index, fila in df_rptaokpt.iterrows():
        numero_doc = str(fila['Número de Documento']).strip()
        lista_documentos = f"31;{numero_doc};AUT-DIAN"

        f.write(lista_documentos + '\n')
```
